# Global Fashion Brands
#### El analisis de datos se enfoca en las marcas de moda que aparecieron en la lista de las 100 mejores marcas mundiales de Interbrand (es una de las consultoras de marca más grandes e importantes del mundo) entre 2001 y 2021.

#### Los datos que se analizarán incluyen:
- **Marca**: El nombre de la marca de moda.
- **País de origen** : El país donde se fundó la marca.
- **Región de origen (América, Europa)**: La región geográfica a la que pertenece la marca.
- **Sector industrial (moda)**: El sector industrial al que pertenece la marca, en este caso, moda.
- **Subsector industrial (confección, cosméticos, artículos de lujo y ropa deportiva)**: El subsector específico dentro de la industria de la moda al que pertenece la marca.

#### Los siguientes campos se extraen para cada año (2001-2021) con valores numéricos:
- **Clasificación de la marca**: La posición de la marca en la lista de las 100 mejores marcas mundiales de Interbrand.
- **Valor de marca (USD billion)**: El valor monetario de la marca en billones de dólares estadounidenses.
- **Crecimiento del valor de marca (%)**: El porcentaje de crecimiento del valor de la marca en comparación con el año anterior.

# Diagnóstico Rápido de Calidad de Datos

```python
# 1. Librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 2. carga de dataset
df_raw = pd.read_excel("FashionBrands.xlsx")
df_before = df_raw.copy()
```

```python
# 3. Mostrar el estado original 
print("Información del dataset original:")
print("Dimensiones:", df_before.shape)
display(df_before.head())
```

```python
# TIPOS DE DATOS 
print("\nTipos de datos:")
display(df_before.dtypes)
```

```python
# VALORES FALTANTES
print("\nValores faltantes:")
display(df_before.isna().sum())
```

```python
# DATOS DUPLICADOS
print("\nDuplicados:", df_before.duplicated().sum())
```

```python
# 4. Copiamos el dataset para trabajar con la limpieza de datos 
df = df_before.copy()
```

# Limpieza de Datos / Tratamiento de Datos 

```python
# 5. Limpiar textos en las columnas 
import re

def norm_text(x):
    if pd.isna(x):
        return x
    return re.sub(r"\s+", " ", str(x)).strip().title()

text_cols = [
    "BrandName",
    "BrandOriginCountry",
    "BrandOriginRegion",
    "BrandSector",
    "BrandSubSector"
]

for col in text_cols:
    df[col] = df[col].apply(norm_text)

print("\nDatos después de limpiar texto:")
print(df[text_cols].head())
```

```python
# 6. Convertir las columnas numéricas correctamente en este caso rank, equity y grow rate
for col in df.columns:
    if col.startswith("Rank") or col.startswith("Equity") or col.startswith("GrowthRate"):
        df[col] = pd.to_numeric(df[col], errors="coerce")

print("\nDespués de convertir columnas numéricas:")
df[["BrandName", "Rank2001", "Equity2001", "GrowthRate2001"]].head(10)

# observacion!!!  el NaN en las columnas numéricas se debe a que había valores no numéricos o vacíos que se convirtieron a NaN al usar pd.to_numeric con errors="coerce". Estos NaN indican que esos registros no tenían un valor numérico válido en esas columnas.
```

```python
# 7. Revisar los valores faltantes
df.isna().sum().sort_values(ascending=False)

# GrowthRate faltantes se manejan con 0 asumiendo que no hubo crecimiento
for col in df.columns:
    if col.startswith("GrowthRate"):
        df[col] = df[col].fillna(0)

# Equity faltantes  mediana por columna
for col in df.columns:
    if col.startswith("Equity"):
        df[col] = df[col].fillna(df[col].median())

print("\nDespués de ver los valores faltantes:")
df.isna().sum().sort_values(ascending=False)
```

### Visualización de las Marcas Dominantes con Mayor Valor en el año 2021 por Región

```python
# Ordenar por Equity2021 de mayor a menor
df_sorted = df.sort_values(by="Equity2021", ascending=False)

# Obtener las 5 mejores marcas por región en 2021
top5_por_region_2021 = (
    df_sorted
    .groupby("BrandOriginRegion")
    .head(5)
    .loc[:, ["BrandOriginRegion", "BrandName", "Equity2021"]]
)

print("Top 5 marcas dominantes por región en 2021:\n")
print(top5_por_region_2021)
```

```python
import matplotlib.pyplot as plt

# Crear figura
plt.figure(figsize=(12, 6))

# Colores pastel
pastel_colors = ["#A8DADC", "#CDB4DB"]

# Graficar por región
for i, (region, data) in enumerate(top5_por_region_2021.groupby("BrandOriginRegion")):
    plt.bar(
        data["BrandName"],
        data["Equity2021"],
        color=pastel_colors[i % len(pastel_colors)],
        label=region
    )

plt.title("Top 5 Marcas Dominantes por Región en 2021\n(Brand Equity en USD Billones)")
plt.ylabel("Valor de Marca 2021 (USD Billones)")
plt.xticks(rotation=45)
plt.legend(title="Región")
plt.tight_layout()
plt.show()
```
